```python
import numpy as np
import pandas as pd
import statistics 

#Part 1: Data Cleaning
df = pd.read_csv('C:\\Lab 4 - Retail_Sales_Data.csv')
from IPython.display import display
display(df)

#This is to fill the missing numeric values in the data with mean
display(df.isnull().sum())
df.fillna(df.select_dtypes(include="number").mean(), inplace=True)  
display(df)

#This is to drop any remaining missing values which have not be been filled
df.dropna(inplace = True)
display(df)

#Ensuring all numeric columns have appropriate data types 
display(df['Quantity'].dtypes)
display(df['Price'].dtypes)
display(df['Total_Spent'].dtypes)
display(df['Discount'].dtypes)

#Converting date column to datetime type
display(df['Date'].dtypes)
df['Date'] = pd.to_datetime(df['Date'])
display(df['Date'].dtypes)

#Part 2: Data Transformation
#Creating a new column Total_Cost calculating total cost of each transaction.
df['Total_Cost']= df['Quantity']*df['Price'] - df['Discount']
display(df)

#o	Creating a new column Day_of_Week that shows the day of the week on which the transaction occurred.
df['Day_of_Week'] = df['Date'].dt.day_name()
display(df)

#Creating a new column Is_Discounted that is True if a discount was applied (Discount > 0), and False otherwise.
df['Is_Discounted'] = df['Discount']>1
display(df)

#Identifying and Handling Outliers for Total_Spent Column.
Q1 = df['Total_Spent'].quantile(0.25)
Q3 = df['Total_Spent'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers_iqr = df[(df['Total_Spent'] < lower_bound) | (df['Total_Spent'] > upper_bound)]
print(f'\nThe outliers for total_spent are:-')
display(outliers_iqr)
df = df[(df['Total_Spent'] >= lower_bound) & (df['Total_Spent'] <= upper_bound)]
display(df)

#Identifying and Handling Outliers for Quantity Column.
Q1 = df['Quantity'].quantile(0.25)
Q3 = df['Quantity'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(f'\nThe outliers for quantity are:-')
outliers_iqr = df[(df['Quantity'] < lower_bound) | (df['Quantity'] > upper_bound)]
display(outliers_iqr)
df = df[(df['Quantity'] >= lower_bound) & (df['Quantity'] <= upper_bound)]
display(df)

#Part 3: Data Analysis
#Desciptive Statistics
# mean_quantity = df['Quantity'].mean().round(2)
# median_quantity = df['Quantity'].median().round(2)
# std_quantity = df['Quantity'].std().round(2)
# var_quantity =df['Quantity'].var().round(2)
# range_quantity= df['Quantity'].max() - df['Quantity'].min()
# print(f"The mean quantity is: {mean_quantity}")
# print(f"The median quantity is: {median_quantity}")
# print(f"The standard deviation of quantity is: {std_quantity}")
# print(f"The variance of quantity is: {var_quantity}")
# print(f"The range of quantity is: {range_quantity}")


# mean_Price = df['Price'].mean().round(2)
# median_Price = df['Price'].median().round(2)
# std_Price = df['Price'].std().round(2)
# var_Price =df['Price'].var().round(2)
# range_Price= (df['Price'].max() - df['Price'].min()).round(2)
# print(f"\nThe mean Price is: {mean_Price}")
# print(f"The median Price is: {median_Price}")
# print(f"The standard deviation of Price is: {std_Price}")
# print(f"The variance of Price is: {var_Price}")
# print(f"The range of Price is: {range_Price}")


# mean_Total_Spent = df['Total_Spent'].mean().round(2)
# median_Total_Spent = df['Total_Spent'].median().round(2)
# std_Total_Spent = df['Total_Spent'].std().round(2)
# var_Total_Spent =df['Total_Spent'].var().round(2)
# range_Total_Spent= (df['Total_Spent'].max() - df['Total_Spent'].min()).round(2)
# print(f"\nThe mean Total_Spent is: {mean_Total_Spent}")
# print(f"The median Total_Spent is: {median_Total_Spent}")
# print(f"The standard deviation of Total_Spent is: {std_Total_Spent}")
# print(f"The variance of Total_Spent is: {var_Total_Spent}")
# print(f"The range of Total_Spent is: {range_Total_Spent}")


# mean_Discount = df['Discount'].mean().round(2)
# median_Discount = df['Discount'].median().round(2)
# std_Discount = df['Discount'].std().round(2)
# var_Discount =df['Discount'].var().round(2)
# range_Discount= (df['Discount'].max() - df['Discount'].min()).round(2)
# print(f"\nThe mean Discount is: {mean_Discount}")
# print(f"The median Discount is: {median_Discount}")
# print(f"The standard deviation of Discount is: {std_Discount}")
# print(f"The variance of Discount is: {var_Discount}")
# print(f"The range of Discount is: {range_Discount}")

# mean_Total_Cost = df['Total_Cost'].mean().round(2)
# median_Total_Cost = df['Total_Cost'].median().round(2)
# std_Total_Cost = df['Total_Cost'].std().round(2)
# var_Total_Cost =df['Total_Cost'].var().round(2)
# range_Total_Cost= df['Total_Cost'].max() - df['Total_Cost'].min()
# print(f"\nThe mean Total_Cost is: {mean_Total_Cost}")
# print(f"The median Total_Cost is: {median_Total_Cost}")
# print(f"The standard deviation of Total_Cost is: {std_Total_Cost}")
# print(f"The variance of Total_Cost is: {var_Total_Cost}")
# print(f"The range of Total_Cost is: {range_Total_Cost}")

display(df.drop(columns=['Date','Transaction_ID','Customer_ID']).describe())


#Transaction Analysis
top_spender = df.groupby('Customer_ID')['Total_Spent'].sum()
print(f'Customer_ID {top_spender.idxmax()} is the customer with the highest total spending.')

popular_product = df.groupby('Product')['Quantity'].sum()
print(f'{popular_product.idxmax()} is the most popular product.')

#Timw-Based Analysis
highest_total_sales = df.groupby('Day_of_Week')['Total_Cost'].sum()
print(f'Highest total sales occurs on {highest_total_sales.idxmax()}.')

lowest_total_sales = df.groupby('Day_of_Week')['Total_Cost'].sum()
print(f'Lowest total sales occurs on {lowest_total_sales.idxmin()}.')

#Correlation Analysis
# numerical_columns = df.select_dtypes(include='number').drop(columns=['Transaction_ID']).corr()
# df[numerical_columns]

numerical_columns = df.drop(columns=['Transaction_ID','Customer_ID']).select_dtypes(include=[np.number]).columns
df[numerical_columns].corr()
```
