```python
import numpy as np
import pandas as pd
from IPython.display import display

# Part 1: Data Cleaning
# Load dataset
df = pd.read_csv('/Users/prabhusana/Downloads/Lab 4 - Retail_Sales_Data.csv')
display(df)

# Check for missing values
display(df.isnull().sum())

# Fill missing numeric values with mean
df.fillna(df.select_dtypes(include="number").mean(), inplace=True)
display(df)

# Drop remaining missing values (non-numeric or not filled)
df.dropna(inplace=True)
display(df)

# Convert Date column to datetime
display(df['Date'].dtypes)
df['Date'].dtype
df['Date'] = pd.to_datetime(df['Date'])

# Check data types of numeric columns
print('The data types of the dataset is:')
df.info()

# Part 2: Data Transformation
# Create Total_Cost column (Quantity * Price - Discount)
df['Total_Cost'] = df['Quantity'] * df['Price'] - df['Discount']
display(df)

# Create Day_of_Week column from Date
df['Day_of_Week'] = df['Date'].dt.day_name()
display(df)

# Create Is_Discounted column (True if Discount > 0)
df['Is_Discounted'] = df['Discount'] > 0
display(df)

# Handle outliers for Total_Spent using IQR
Q1 = df['Total_Spent'].quantile(0.25)
Q3 = df['Total_Spent'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers_iqr = df[(df['Total_Spent'] < lower_bound) | (df['Total_Spent'] > upper_bound)]
print('\nThe outliers for Total_Spent are:')
display(outliers_iqr)
df = df[(df['Total_Spent'] >= lower_bound) & (df['Total_Spent'] <= upper_bound)]
display(df)


Q1 = df['Total_Spent'].quantile(0.25)
Q3 = df['Total_Spent'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers_iqr = df[(df['Total_Spent'] < lower_bound) | (df['Total_Spent'] > upper_bound)]
print('\nThe outliers for Total_Spent are:')
display(outliers_iqr)
df = df[(df['Total_Spent'] >= lower_bound) & (df['Total_Spent'] <= upper_bound)]
display(df)

# Handle outliers for Quantity using IQR

Q1 = df['Quantity'].quantile(0.25)
Q3 = df['Quantity'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print('\nThe outliers for Quantity are:')
outliers_iqr = df[(df['Quantity'] < lower_bound) | (df['Quantity'] > upper_bound)]
display(outliers_iqr)
df = df[(df['Quantity'] >= lower_bound) & (df['Quantity'] <= upper_bound)]
display(df)

# Part 3: Data Analysis
# Descriptive statistics (excluding non-relevant columns)
display(df.drop(columns=['Date', 'Transaction_ID', 'Customer_ID', 'Product', 'Payment_Method', 'Day_of_Week', 'Is_Discounted']).describe())

# Transaction Analysis
top_spender = df.groupby('Customer_ID')['Total_Spent'].sum()
print(f'Customer_ID {top_spender.idxmax()} is the customer with the highest total spending.')

popular_product = df.groupby('Product')['Quantity'].sum()
print(f'{popular_product.idxmax()} is the most popular product.')

# Time-Based Analysis
highest_total_sales = df.groupby('Day_of_Week')['Total_Cost'].sum()
print(f'Highest total sales occurs on {highest_total_sales.idxmax()}.')

lowest_total_sales = df.groupby('Day_of_Week')['Total_Cost'].sum()
print(f'Lowest total sales occurs on {lowest_total_sales.idxmin()}.')

# Correlation Analysis
numerical_columns = df.drop(columns=['Transaction_ID', 'Customer_ID', 'Date', 'Product', 'Payment_Method', 'Day_of_Week', 'Is_Discounted']).select_dtypes(include=[np.number]).columns
correlation_matrix = df[numerical_columns].corr()
display(correlation_matrix)
```
