```python
#RAMANA PRABHU SANA - rsana@purdue.edu
#SIDDHARTH SHENOY - shenoy22@purdue.edu

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
import scipy.stats as stats

# Data Source - https://www.kaggle.com/datasets/ahmadrazakashif/bmw-worldwide-sales-records-20102024

# Load the dataset
# Make sure 'BMW sales data (2010-2024) (1).csv' is in the same directory as your notebook
df=pd.read_csv('/Users/prabhusana/Downloads/BMW sales data (2010-2024) (1).csv')

#
# 2. Data Cleaning: 
#
#This is to fill the missing numeric values in the data with mean
display(df.isnull().sum())
df.fillna(df.select_dtypes(include="number").mean(), inplace=True)  
# display(df)

#This is to drop any remaining missing values which have not be been filled
df.dropna(inplace = True)
print("data after dropping any data (if applicable):-")
display(df)

# Checked for duplicate values in the DataFrame.
display(df.duplicated().sum())
df.drop_duplicates()


#Identifying and Handling Outliers

# --- Outlier Removal for Price_USD Column (Iterative) ---
print(f"Initial rows: {len(df)}")
print("\n--- Cleaning Price_USD ---\n")
while True:
    Q1 = df['Price_USD'].quantile(0.25)
    Q3 = df['Price_USD'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Identify outliers
    outliers_iqr = df[(df['Price_USD'] < lower_bound) | (df['Price_USD'] > upper_bound)]
    num_outliers = len(outliers_iqr)
    
    # Break loop if no outliers are found
    if num_outliers == 0:
        print("No outliers found for Price_USD.\n")
        break
    
    # Print the number found and remove them
    print(f'Found and removing {num_outliers} outliers for Price_USD...')
    df = df[(df['Price_USD'] >= lower_bound) & (df['Price_USD'] <= upper_bound)]
    print(f"Rows remaining: {len(df)}")

display(df)

# --- Outlier Removal for Sales_Volume Column (Iterative) ---
print("\n--- Cleaning Sales_Volume ---\n")
while True:
    Q1 = df['Sales_Volume'].quantile(0.25)
    Q3 = df['Sales_Volume'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Identify outliers
    outliers_iqr = df[(df['Sales_Volume'] < lower_bound) | (df['Sales_Volume'] > upper_bound)]
    num_outliers = len(outliers_iqr)
    
    # Break loop if no outliers are found
    if num_outliers == 0:
        print("No outliers found for Sales_Volume.\n")
        break
    
    # Print the number found and remove them
    print(f'Found and removing {num_outliers} outliers for Sales_Volume...')
    df = df[(df['Sales_Volume'] >= lower_bound) & (df['Sales_Volume'] <= upper_bound)]
    print(f"Rows remaining: {len(df)}")

display(df)

# --- Outlier Removal for Mileage_KM (Iterative) ---
print("\n--- Cleaning Mileage_KM ---\n")
while True:
    Q1 = df['Mileage_KM'].quantile(0.25)
    Q3 = df['Mileage_KM'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Identify outliers
    outliers_iqr = df[(df['Mileage_KM'] < lower_bound) | (df['Mileage_KM'] > upper_bound)]
    num_outliers = len(outliers_iqr)
    
    # Break loop if no outliers are found
    if num_outliers == 0:
        print("No outliers found for Mileage_KM.\n")
        break
    
    # Print the number found and remove them
    print(f'Found and removing {num_outliers} outliers for Mileage_KM...')
    df = df[(df['Mileage_KM'] >= lower_bound) & (df['Mileage_KM'] <= upper_bound)]
    print(f"Rows remaining: {len(df)}")

display(df)

# --- Outlier Removal for Engine_Size_L (Iterative) ---
print("\n--- Cleaning Engine_Size_L ---\n")
while True:
    Q1 = df['Engine_Size_L'].quantile(0.25)
    Q3 = df['Engine_Size_L'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Identify outliers
    outliers_iqr = df[(df['Engine_Size_L'] < lower_bound) | (df['Engine_Size_L'] > upper_bound)]
    num_outliers = len(outliers_iqr)
    
    # Break loop if no outliers are found
    if num_outliers == 0:
        print("No outliers found for Engine_Size_L.\n")
        break
    
    # Print the number found and remove them
    print(f'Found and removing {num_outliers} outliers for Engine_Size_L...')
    df = df[(df['Engine_Size_L'] >= lower_bound) & (df['Engine_Size_L'] <= upper_bound)]
    print(f"Rows remaining: {len(df)}")

display(df)

# Display the final, cleaned DataFrame after all loops are complete
print("\n--- Final DataFrame after all outlier removal ---")
display(df)

# 3. Exploratory Data Analysis (EDA): 
#Descriptive Statistics
display(df.drop(columns=['Year','Engine_Size_L']).describe())


# Data Visualization
#Boxplot: Volume of Sales By Model
print("We used boxplots to compare spread, median, and outliers of units sold across various BMW models.\n"
    "Compared to bar charts, boxplots provide much richer statistical summary and directly reveal spread and variation between models,\n"
      "which are essential for understanding business dynamism among categories.")
plt.figure(figsize=(18,12))
sns.set_style('darkgrid')
sns.boxplot(x=df['Model'],y= df['Sales_Volume'],hue=df['Model'])
plt.xlabel('Model Sold')
plt.ylabel('Volume of Sales')
plt.title('Volume of Sales By Model')
plt.xticks(rotation =45)
plt.show()


#Countplot: Fuel_Type vs Transmission
print("We used countplot to compare fuel type vs transmission.\n"
      "We used countplot as it is ideal to show frequency of each category.")
plt.figure(figsize=(14,10))
sns.set_style('darkgrid')
ax = sns.countplot(x=df['Fuel_Type'], hue=df['Transmission'])

plt.title("Fuel_Type vs Transmission")
plt.xlabel("Fuel_Type")
plt.ylabel("Vehicle Count")

# Add count labels on bars
for p in ax.patches:
    height = p.get_height()
    ax.text(p.get_x() + p.get_width()/2., height + 50, int(height), ha='center', fontsize=12)

# Legend inside grid, above bars, away from labels
plt.legend(title='Transmission', loc='upper left', bbox_to_anchor=(1, 1), fontsize=12, title_fontsize=13, frameon=True)

plt.tight_layout()
plt.show()


#Bar Chart: Distribution of Sales Records by Region
print("we used bar plot to showcase which region has highest sales records.")
plt.figure(figsize=(10,6))
counts = df['Region'].value_counts()
ax = sns.barplot(x=counts.values, y=counts.index, hue = counts.index)

# Add labels to bars
for i, v in enumerate(counts.values):
    ax.text(v + 5, i, str(v), color='black', va='center')

plt.title('Sales Records Distribution by Region')
plt.xlabel('Count')
plt.ylabel('Region')
plt.show()

#Boxplot: Price Distribution by Region
print("We used boxplot to showcase price across regions to understand about the pricing dynamics.")
plt.figure(figsize=(14,8))
sns.boxplot(x='Region', y='Price_USD', hue = df['Region'],data=df)
plt.title('Price Distribution by Region')
plt.xlabel('Region')
plt.ylabel('Price (USD)')
plt.xticks(rotation=45)
plt.show()

# Correlation Analysis
print("Correlation Analysis:-")
print("We used heatmap to show correlation to visualize pairwise relationships and strength between all continuous variables at once.\n"
      "A heatmap quickly shows all mutual correlations compared to other visualizations and helps to guide further analysis or modeling.")
correlation = df[["Engine_Size_L", "Mileage_KM", "Price_USD", "Sales_Volume"]].corr()
print("\nCorrelation matrix:")
display(correlation)
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

# Hypothesis Testing
# FuelType vs AveragePrice
print("Hypothesis Testing:-")
print("Null Hypothesis - H₀: There is no significant difference in average vehicle price between fuel types (Petrol, Diesel, Electric).\n"
    "Alternative Hypothesis - Ha: At least one fuel type has a different average price.")
groups = [y["Price_USD"].values for x, y in df.groupby("Fuel_Type")]
stat, p = stats.f_oneway(*groups)
print("ANOVA for Price by Fuel_Type: p-value =", p)
print("Testing at 95% significance level.")
if p < 0.05:
    print("Conclusion: Significant difference in mean prices among fuel types.")
else:
    print("Conclusion: No significant difference in mean prices among fuel types.")

# Region vs AveragePrice
print("\n Null Hypothesis - H₀: The average vehicle price is the same across all regions.\n"
    "Alternative Hypothesis - Ha: At least one region differs in mean price.")
groups = [y["Price_USD"].values for x, y in df.groupby("Region")]
stat, p = stats.f_oneway(*groups)
print("ANOVA for Price by Region: p-value =", p)
print("Testing at 95% significance level.")
if p < 0.05:
    print("Conclusion:At least one region differs in mean price.")
else:
    print("Conclusion: The average vehicle price is the same across all regions.")

# FuelType vs Transmission
print("\n Null Hypothesis - H₀: Fuel type and transmission are independent.\n"
    " Alternative Hypothesis - Ha: Fuel type and transmission are associated.")
contingency = pd.crosstab(df['Fuel_Type'], df['Transmission'])
chi2, p, dof, ex = stats.chi2_contingency(contingency)
print("Chi-squared p-value for Fuel_Type vs Transmission:", p)
print("Testing at 95% significance level.")
if  p < 0.05:
    print("Conclusion: Fuel type and transmission are associated.")
else:
    print("Conclusion: No association found between fuel type and transmission.")

#Time series Analysis
print("\nWe used lineplot to show the trend of sales over time.\n"
     "Ideal for trends and time-series analysis,\n"
      "a lineplot reveals cyclical patterns and long-term changes, which would be missed with a bar chart or point plot.")
plt.figure(figsize=(18,12))
sns.set_style('darkgrid')
sns.lineplot(x = df['Year'],y= df['Sales_Volume'])
plt.xlabel('Year')
plt.ylabel('Sales Volume')
plt.title('Sales Over Time')
plt.show()

#
# 6. Extracting and Summarizing Business Insights
#
print("\nBusiness Insights Summary:")
print("- Visual analysis shows significant price and sales variation by region and model for Petrol and Diesel models (see boxplots).")
print("- Statistically, there is **no significant difference** in average price by region (p=0.426) or by fuel type (p=0.53), as shown by the ANOVA tests.")
print("- Similarly, there is **no statistical association** between fuel type and transmission (p=0.659), as per the Chi-squared test.")
print("- Highest sales volumes are seen in select models and regions (as shown in the 'Volume of Sales By Model' plot).")
print("- There is clear time-based cyclicality and recent growth trends in the product line (as shown in the 'Sales Over Time' plot).")

print("\nRecommendations:")
print("- Focus on promoting top-selling models in regions that show high sales volumes and recent growth.")
print("- Despite the lack of *statistical* significance, the *visual* differences in price medians and spreads between regions (from the boxplot) suggest that targeted regional pricing strategies could still be tested to optimize profits.")
print("- Monitor the cyclical sales patterns identified in the time-series analysis to manage inventory and plan promotions.")
```

```python
# RAMANA PRABHU SANA - rsana@purdue.edu
# SIDDHARTH SHENOY - shenoy22@purdue.edu

# Import necessary libraries
import matplotlib.pyplot as plt  # For plotting
import pandas as pd  # For data manipulation
import numpy as np  # For numerical operations
import seaborn as sns  # For advanced visualization
import scipy.stats as stats  # For statistical tests

# Data Source - https://www.kaggle.com/datasets/ahmadrazakashif/bmw-worldwide-sales-records-20102024

# Load the dataset
# Make sure 'BMW sales data (2010-2024) (1).csv' is in the same directory as your notebook
# Note: This is a local path and will only work on your machine.
df=pd.read_csv('/Users/prabhusana/Downloads/BMW sales data (2010-2024) (1).csv')


# 2. Data Cleaning: 
#
# Check for missing (null) values and display the count for each column
display(df.isnull().sum())
# Fill any missing values in numeric columns with the mean of that column
df.fillna(df.select_dtypes(include="number").mean(), inplace=True)
# display(df) # Original commented-out display

# This is to drop any remaining missing values which have not be been filled (e.g., in non-numeric columns)
df.dropna(inplace = True)
print("data after dropping any data (if applicable):-")
display(df) # Show the DataFrame after dropping remaining nulls

# Checked for duplicate values in the DataFrame and display the total count
display(df.duplicated().sum())
# Remove all duplicate rows from the DataFrame
df.drop_duplicates() # Note: This line as-is doesn't modify df. You might intend: df = df.drop_duplicates()


# Identifying and Handling Outliers

# --- Outlier Removal for Price_USD Column (Iterative) ---
print(f"Initial rows: {len(df)}")
print("\n--- Cleaning Price_USD ---\n")
# Start an iterative loop to remove outliers
while True:
    # Calculate Q1 (25th percentile) and Q3 (75th percentile)
    Q1 = df['Price_USD'].quantile(0.25)
    Q3 = df['Price_USD'].quantile(0.75)
    # Calculate the Interquartile Range (IQR)
    IQR = Q3 - Q1
    # Define the lower and upper bounds for outlier detection
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Identify outliers based on the bounds
    outliers_iqr = df[(df['Price_USD'] < lower_bound) | (df['Price_USD'] > upper_bound)]
    num_outliers = len(outliers_iqr) # Count the number of outliers found
    
    # Break loop if no outliers are found
    if num_outliers == 0:
        print("No outliers found for Price_USD.\n")
        break # Exit the loop
    
    # Print the number found and remove them
    print(f'Found and removing {num_outliers} outliers for Price_USD...')
    # Keep only the rows that are *within* the bounds (not outliers)
    df = df[(df['Price_USD'] >= lower_bound) & (df['Price_USD'] <= upper_bound)]
    print(f"Rows remaining: {len(df)}") # Print remaining row count

display(df) # Show DataFrame after cleaning Price_USD

# --- Outlier Removal for Sales_Volume Column (Iterative) ---
print("\n--- Cleaning Sales_Volume ---\n")
# Repeat the same outlier removal process for 'Sales_Volume'
while True:
    Q1 = df['Sales_Volume'].quantile(0.25)
    Q3 = df['Sales_Volume'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Identify outliers
    outliers_iqr = df[(df['Sales_Volume'] < lower_bound) | (df['Sales_Volume'] > upper_bound)]
    num_outliers = len(outliers_iqr)
    
    # Break loop if no outliers are found
    if num_outliers == 0:
        print("No outliers found for Sales_Volume.\n")
        break
    
    # Print the number found and remove them
    print(f'Found and removing {num_outliers} outliers for Sales_Volume...')
    df = df[(df['Sales_Volume'] >= lower_bound) & (df['Sales_Volume'] <= upper_bound)]
    print(f"Rows remaining: {len(df)}")

display(df) # Show DataFrame after cleaning Sales_Volume

# --- Outlier Removal for Mileage_KM (Iterative) ---
print("\n--- Cleaning Mileage_KM ---\n")
# Repeat the same outlier removal process for 'Mileage_KM'
while True:
    Q1 = df['Mileage_KM'].quantile(0.25)
    Q3 = df['Mileage_KM'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Identify outliers
    outliers_iqr = df[(df['Mileage_KM'] < lower_bound) | (df['Mileage_KM'] > upper_bound)]
    num_outliers = len(outliers_iqr)
    
    # Break loop if no outliers are found
    if num_outliers == 0:
        print("No outliers found for Mileage_KM.\n")
        break
    
    # Print the number found and remove them
    print(f'Found and removing {num_outliers} outliers for Mileage_KM...')
    df = df[(df['Mileage_KM'] >= lower_bound) & (df['Mileage_KM'] <= upper_bound)]
    print(f"Rows remaining: {len(df)}")

display(df) # Show DataFrame after cleaning Mileage_KM

# --- Outlier Removal for Engine_Size_L (Iterative) ---
print("\n--- Cleaning Engine_Size_L ---\n")
# Repeat the same outlier removal process for 'Engine_Size_L'
while True:
    Q1 = df['Engine_Size_L'].quantile(0.25)
    Q3 = df['Engine_Size_L'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Identify outliers
    outliers_iqr = df[(df['Engine_Size_L'] < lower_bound) | (df['Engine_Size_L'] > upper_bound)]
    num_outliers = len(outliers_iqr)
    
    # Break loop if no outliers are found
    if num_outliers == 0:
        print("No outliers found for Engine_Size_L.\n")
        break
    
    # Print the number found and remove them
    print(f'Found and removing {num_outliers} outliers for Engine_Size_L...')
    df = df[(df['Engine_Size_L'] >= lower_bound) & (df['Engine_Size_L'] <= upper_bound)]
    print(f"Rows remaining: {len(df)}")

display(df) # Show DataFrame after cleaning Engine_Size_L

# Display the final, cleaned DataFrame after all loops are complete
print("\n--- Final DataFrame after all outlier removal ---")
display(df)

# 3. Exploratory Data Analysis (EDA): 
# Descriptive Statistics
# Generate summary statistics (mean, std, min, max, quartiles) for numeric columns
# Drop 'Year' and 'Engine_Size_L' from this specific summary view
display(df.drop(columns=['Year','Engine_Size_L']).describe())


# Data Visualization
# Boxplot: Volume of Sales By Model
# Print the justification for using this plot type
print("We used boxplots to compare spread, median, and outliers of units sold across various BMW models.\n")
print("Compared to bar charts, boxplots provide much richer statistical summary and directly reveal spread and variation between models which are essential for understanding business dynamism among categories.\n")
plt.figure(figsize=(18,12)) # Set the figure size
sns.set_style('darkgrid') # Set the plot style
# Create a boxplot
sns.boxplot(x=df['Model'],y= df['Sales_Volume'],hue=df['Model'])
# Set labels and title
plt.xlabel('Model Sold')
plt.ylabel('Volume of Sales')
plt.title('Volume of Sales By Model')
plt.xticks(rotation =45) # Rotate x-axis labels for readability
plt.show() # Display the plot


# Countplot: Fuel_Type vs Transmission
# Print the justification for using this plot type
print("We used countplot to compare fuel type vs transmission.We used countplot as it is ideal to show frequency of each category.")
plt.figure(figsize=(14,10)) # Set figure size
sns.set_style('darkgrid') # Set style
# Create a countplot with 'Transmission' as the hue
ax = sns.countplot(x=df['Fuel_Type'], hue=df['Transmission'])

# Set title and labels
plt.title("Fuel_Type vs Transmission")
plt.xlabel("Fuel_Type")
plt.ylabel("Vehicle Count")

# Add count labels on bars
# Loop through each bar ('patch') in the plot
for p in ax.patches:
    height = p.get_height() # Get the height of the bar
    # Add text at the top of the bar
    ax.text(p.get_x() + p.get_width()/2., height + 50, int(height), ha='center', fontsize=12)

# Legend inside grid, above bars, away from labels
# Place the legend outside the top-right corner of the plot
plt.legend(title='Transmission', loc='upper left', bbox_to_anchor=(1, 1), fontsize=12, title_fontsize=13, frameon=True)

plt.tight_layout() # Adjust plot to prevent label overlap
plt.show() # Display the plot


# Bar Chart: Distribution of Sales Records by Region
# Print the justification for using this plot type
print("we used bar plot to showcase which region has highest sales records.")
plt.figure(figsize=(10,6)) # Set figure size
# Get the counts of each 'Region'
counts = df['Region'].value_counts()
# Create a horizontal bar plot (y=index, x=values)
ax = sns.barplot(x=counts.values, y=counts.index, hue = counts.index)

# Add labels to bars
# Loop through the bars and add text labels
for i, v in enumerate(counts.values):
    ax.text(v + 5, i, str(v), color='black', va='center') # Add text at (value + 5, index)

# Set title and labels
plt.title('Sales Records Distribution by Region')
plt.xlabel('Count')
plt.ylabel('Region')
plt.show() # Display the plot

# Boxplot: Price Distribution by Region
# Print the justification for using this plot type
print("We used boxplot to showcase price across regions to understand about the pricing dynamics.")
plt.figure(figsize=(14,8)) # Set figure size
# Create a boxplot of 'Price_USD' grouped by 'Region'
sns.boxplot(x='Region', y='Price_USD', hue = df['Region'],data=df)
# Set title and labels
plt.title('Price Distribution by Region')
plt.xlabel('Region')
plt.ylabel('Price (USD)')
plt.xticks(rotation=45) # Rotate x-axis labels
plt.show() # Display the plot

# Correlation Analysis
print("Correlation Analysis:-")
# Print the justification for using this plot type
print("We used heatmap to show correlation to visualize pairwise relationships and strength between all continuous variables at once.\n"
      "A heatmap quickly shows all mutual correlations compared to other visualizations and helps to guide further analysis or modeling.")
# Calculate the correlation matrix for the specified numeric columns
correlation = df[["Engine_Size_L", "Mileage_KM", "Price_USD", "Sales_Volume"]].corr()
print("\nCorrelation matrix:")
display(correlation) # Display the correlation matrix as a table
plt.figure(figsize=(8, 6)) # Set figure size
# Create the heatmap
# annot=True adds the numeric values to each cell
# cmap='coolwarm' sets the color scheme
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap") # Set title
plt.show() # Display the plot

# Hypothesis Testing
# FuelType vs AveragePrice (ANOVA)
print("Hypothesis Testing:-")
# State the null and alternative hypotheses
print("Null Hypothesis - H₀: There is no significant difference in average vehicle price between fuel types (Petrol, Diesel, Electric).\n"
  "Alternative Hypothesis - Ha: At least one fuel type has a different average price.")
# Create a list of price arrays, one for each fuel type
groups = [y["Price_USD"].values for x, y in df.groupby("Fuel_Type")]
# Perform the one-way ANOVA test
stat, p = stats.f_oneway(*groups)
print("ANOVA for Price by Fuel_Type: p-value =", p) # Print the p-value
print("Testing at 95% significance level.")
# Check if the p-value is less than the significance level (0.05)
if p < 0.05:
    print("Conclusion: Significant difference in mean prices among fuel types.")
else:
    print("Conclusion: No significant difference in mean prices among fuel types.")

# Region vs AveragePrice (ANOVA)
# State the null and alternative hypotheses
print("\n Null Hypothesis - H₀: The average vehicle price is the same across all regions.\n"
  "Alternative Hypothesis - Ha: At least one region differs in mean price.")
# Create a list of price arrays, one for each region
groups = [y["Price_USD"].values for x, y in df.groupby("Region")]
# Perform the one-way ANOVA test
stat, p = stats.f_oneway(*groups)
print("ANOVA for Price by Region: p-value =", p) # Print the p-value
print("Testing at 95% significance level.")
# Check the p-value
if p < 0.05:
    print("Conclusion:At least one region differs in mean price.")
else:
    print("Conclusion: The average vehicle price is the same across all regions.")

# FuelType vs Transmission (Chi-squared)
# State the null and alternative hypotheses
print("\n Null Hypothesis - H₀: Fuel type and transmission are independent.\n"
  "Alternative Hypothesis - Ha: Fuel type and transmission are associated.")
# Create a contingency table (crosstab) of counts
contingency = pd.crosstab(df['Fuel_Type'], df['Transmission'])
# Perform the Chi-squared test for independence
chi2, p, dof, ex = stats.chi2_contingency(contingency)
print("Chi-squared p-value for Fuel_Type vs Transmission:", p) # Print the p-value
print("Testing at 95% significance level.")
# Check the p-value
if  p < 0.05:
    print("Conclusion: Fuel type and transmission are associated.")
else:
    print("Conclusion: No association found between fuel type and transmission.")

# Time series Analysis
# Print the justification for using this plot type
print("\nWe used lineplot to show the trend of sales over time.\n"
  "Ideal for trends and time-series analysis. A lineplot reveals cyclical patterns and long-term changes, which would be missed with a bar chart or point plot.")
plt.figure(figsize=(18,12)) # Set figure size
sns.set_style('darkgrid') # Set style
# Create a line plot of 'Sales_Volume' over 'Year'
sns.lineplot(x = df['Year'],y= df['Sales_Volume'])
# Set labels and title
plt.xlabel('Year')
plt.ylabel('Sales Volume')
plt.title('Sales Over Time')
plt.show() # Display the plot

#
# 6. Extracting and Summarizing Business Insights
#
# Print the final conclusions from the analysis
print("\nBusiness Insights Summary:")
print("- Visual analysis shows significant price and sales variation by region and model for Petrol and Diesel models (see boxplots).")
print("- Statistically, there is **no significant difference** in average price by region (p=0.426) or by fuel type (p=0.53), as shown by the ANOVA tests.")
print("- Similarly, there is **no statistical association** between fuel type and transmission (p=0.659), as per the Chi-squared test.")
print("- Highest sales volumes are seen in select models and regions (as shown in the 'Volume of Sales By Model' plot).")
print("- There is clear time-based cyclicality and recent growth trends in the product line (as shown in the 'Sales Over Time' plot).")

# Print the actionable recommendations based on the insights
print("\nRecommendations:")
print("- Focus on promoting top-selling models in regions that show high sales volumes and recent growth.")
print("- Despite the lack of *statistical* significance, the *visual* differences in price medians and spreads between regions (from the boxplot) suggest that targeted regional pricing strategies could still be tested to optimize profits.")
print("- Monitor the cyclical sales patterns identified in the time-series analysis to manage inventory and plan promotions.")
```
