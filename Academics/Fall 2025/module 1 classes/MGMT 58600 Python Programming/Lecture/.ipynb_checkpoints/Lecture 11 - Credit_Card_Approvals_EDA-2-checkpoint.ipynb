# Credit Card Approvals - Exploratory Data Analysis
This notebook performs an exploratory data analysis (EDA) on the credit card approvals dataset using only NumPy and Pandas.

## Load the Data
Reading the dataset into a Pandas DataFrame.

```python
import pandas as pd
import numpy as np

# Load the dataset
file_path = "cc_approvals.csv"
df = pd.read_csv(file_path)

# Display first five rows
df.head()
```

```python
# Display last five rows
df.tail()
```

```python
# Display a random sample of rows
df.sample(5)
```

## Initial Inspection
Checking the dataset structure, column names, and data types.

```python
# Basic dataset information
df.info()
```

## Missing Values Analysis
Identifying any missing data in the dataset.

```python
# Checking for missing values
df.isnull().sum()
```

## Basic Statistical Summary
Generating summary statistics for numerical and categorical features.

```python
# Basic statistical summary for numerical features
df.describe()
```

```python
# Summary for categorical features
df.describe(include=[object])
```

## Duplicate Detection
Checking if there are any duplicate records in the dataset.

```python
# Checking for duplicate rows
print(df.duplicated().sum())
```

## Feature Distribution
Understanding the uniqueness of different features.

```python
# Checking the number of unique values in each column
df.nunique()
```

## Correlation Analysis
Computing correlations between numerical features.

```python
# Correlation analysis for numerical features
numerical_columns = df.select_dtypes(include=[np.number]).columns
df[numerical_columns].corr()
```

## Business Insights
Answering key analytical questions from the dataset.

  1. What is the average income of approved vs non-approved applicants?
  2. How does prior default history impact approval rates?
  3. What percentage of applicants have been employed for more than 5 years?
  4. What is the relationship between credit score and approval?
  5. Are married applicants more likely to get approved?
  6. Do applicants with higher debt levels get approved less often?
  7. How does employment status impact approval rates?
  8. What percentage of applicants have a driver’s license?


```python
# Analytical Questions & Insights

# 1. What is the average income of approved vs non-approved applicants?
income_approved = df[df['Approved'] == 1]['Income'].mean()
income_not_approved = df[df['Approved'] == 0]['Income'].mean()
print(f"Approved applicants have an average income of {income_approved:.2f}, whereas non-approved applicants have {income_not_approved:.2f}")
```

```python
# 2. How does prior default history impact approval rates?
prior_default_approval_rate = df.groupby('PriorDefault')['Approved'].mean()
print(f"Applicants with prior defaults have an approval rate of:\n{prior_default_approval_rate}")
```

```python
# 3. What percentage of applicants have been employed for more than 5 years?
employed_more_than_5_years = (df[df['YearsEmployed'] > 5].shape[0] / df.shape[0]) * 100
print(f"{employed_more_than_5_years:.2f}% of applicants have been employed for more than 5 years.")
```

```python
# 4. What is the relationship between credit score and approval?
credit_score_approval = df.groupby('Approved')['CreditScore'].mean()
print(f"Average credit score of approved applicants: {credit_score_approval[1]:.2f}, non-approved: {credit_score_approval[0]:.2f}.")
```

```python
# 5. Are married applicants more likely to get approved?
married_approval_rate = df.groupby('Married')['Approved'].mean()
print(f"Approval rate by marital status:\n{married_approval_rate.to_dict()}")
```

```python
# 6. Do applicants with higher debt levels get approved less often?
debt_approval_correlation = df[['Debt', 'Approved']].corr().iloc[0,1]
print(f"Debt level has a correlation of {debt_approval_correlation:.2f} with approval")
```

```python
average_debt_by_approval = df.groupby('Approved')['Debt'].mean()
print(f'Average debt of approved applicants: {average_debt_by_approval[1]:.2f} vs. non-approved: {average_debt_by_approval[0]:.2f}')
```

```python
# 7. How does employment status impact approval rates?
employed_approval_rate = df.groupby('Employed')['Approved'].mean()
print(f"Approval rate by employment status:\n{employed_approval_rate}")
```

```python
# 8. What percentage of applicants have a driver’s license?
drivers_license_percentage = (df['DriversLicense'].sum() / df.shape[0]) * 100
print(f"{drivers_license_percentage:.2f}% of applicants have a driver’s license.")
```
